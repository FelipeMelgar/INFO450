#For Loops

```python
fruits=["apple","banana", "grape"]
```

```python
for fruit in fruits:
  print(fruits)
```

```python
numbers=[10,20,30,40]
total=0
count=0
#I want to use a loop to find the total of numbers
for number in numbers:
  total=total+number
  count=count+1
  print(f"This is iteration {count} and the total is {total}.")
print(total)
```

```python
numbers=[3,5,7,2,8,1]
max_number=0
for number in numbers:
  if number>max_number:
    max_number=number
    print(number)
print(f"The biggest number is {max_number}.")
```

#Iterating over a string

```python
words="banana"
for letter in words:
  print(letter)
```

```python
#how many a's in the word
word="banana"
count=0
for letter in word:
  if letter=="a":
    count=count+1
print(f"The letter a is used {count} times")
```

```python
even_numbers=[]#makes an empty list

for i in range(1,21):
  if i %2==0:
    even_numbers.append(i)
even_numbers
```

#While loops

```python
data=["yes","yes","no",1,"no",0,"yes"]
clean_data=[]
while len(data)>0: #continue as long as list has numbers
  entry=data.pop(0)
  if entry !=0:
    clean_data.append(entry)
clean_data
```
